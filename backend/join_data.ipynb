# Join traffic with pollution data

```python
import pandas as pd
```

```python
traffic = pd.read_csv("database\\traffic_date_value.csv")
pollution = pd.read_csv("database\\pollution_date_value.csv")
```

```python
df_final = traffic.merge(pollution, left_on='FECHA_HORA', right_on='FECHA_HORA', how='left')
```

```python
df_final
```

```python
df_final = df_final.drop(columns=['MAGNITUD', 'TECNICA', 'CODIGO_ESTACION', 'NOMBRE_ESTACION', 'COD_MAGNITUD', 
                 'NOMBRE_MAGNITUD', 'ABREVIATURA', 'UNIDAD', 'COD_TECNICA',	'TECNICA_MEDIDA'])
```

```python
df_final.info()
```

```python
df_final = df_final.dropna()
```
