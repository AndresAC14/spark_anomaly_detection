# Train model

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import xgboost as xgb
from sklearn.metrics import root_mean_squared_error
```

```python
# Suponiendo que df_final es tu DataFrame combinado
df = df_final.copy()
```

```python
df['FECHA_HORA'] = pd.to_datetime(df['FECHA_HORA'])
```

```python
# Extraer características temporales
df['hora'] = df['FECHA_HORA'].dt.hour
df['dia_semana'] = df['FECHA_HORA'].dt.dayofweek
df['mes'] = df['FECHA_HORA'].dt.month
```

```python
# Codificar IDELEM (puedes probar con OneHot más adelante)
le = LabelEncoder()
df['idelem_encoded'] = le.fit_transform(df['IDELEM'])
```

```python
# Definir variables
features = ['idelem_encoded', 'INTENSIDAD', 'OCUPACION', 'CARGA', 'hora', 'dia_semana', 'mes']
target = 'VALOR_CONTAMINACION'
```

```python
# Separar en X e y
X = df[features]
y = df[target]
```

```python
# Dividir entrenamiento y test por tiempo (opcionalmente por FECHA_HORA)
X_train, X_test, y_train, y_test = train_test_split(X, y, shuffle=False, test_size=0.2)
```

```python
import numpy as np

# Revisión rápida
print("NaNs en y:", y.isna().sum())
print("Inf en y:", np.isinf(y).sum())
print("Valores extremos:", y.max(), y.min())
```

```python
# Entrenar modelo
model = xgb.XGBRegressor(objective='reg:squarederror')
model.fit(X_train, y_train)
```

```python
# Evaluar
y_pred = model.predict(X_test)
rmse = root_mean_squared_error(y_test, y_pred)
print(f"RMSE: {rmse:.2f}")
```

```python
# Asegúrate de que FECHA_HORA sea datetime
df['FECHA_HORA'] = pd.to_datetime(df['FECHA_HORA'])

# Extrae la hora para usar como feature
df['HORA'] = df['FECHA_HORA'].dt.hour

# Agrupa por IDELEM y FECHA_HORA para asegurarte de que esté bien resampleado
df = df.sort_values(['IDELEM', 'FECHA_HORA'])
```

```python
features = ['HORA', 'IDELEM', 'INTENSIDAD', 'OCUPACION', 'CARGA']
target = 'VALOR_CONTAMINACION'

# Convertimos IDELEM a categoría para usar en modelos
df['IDELEM'] = df['IDELEM'].astype('category')
```

```python
train = df[df['FECHA_HORA'] < '2024-01-01']
test = df[df['FECHA_HORA'] >= '2024-01-01']
```

```python
from lightgbm import LGBMRegressor
from sklearn.metrics import mean_squared_error

X_train = train[features]
y_train = train[target]
X_test = test[features]
y_test = test[target]

model = LGBMRegressor()
model.fit(X_train, y_train)

# Predicción
y_pred = model.predict(X_test)
print("RMSE:", mean_squared_error(y_test, y_pred))
```

```python
from datetime import datetime

# Simulamos condiciones actuales (puedes usar promedios, último valor, etc.)
hora_actual = pd.Timestamp.now().replace(minute=0, second=0, microsecond=0)

# Creamos predicciones por hora del día para hoy
horas = list(range(24))
idelems = df['IDELEM'].unique()

future_df = pd.DataFrame([(h, idelem) for h in horas for idelem in idelems], columns=['HORA', 'IDELEM'])

# Añadir features sintéticos. Puedes usar medias históricas o valores recientes
# Aquí simplemente usamos la media por IDELEM
agg = df.groupby('IDELEM')[['INTENSIDAD', 'OCUPACION', 'CARGA']].mean().reset_index()
future_df = future_df.merge(agg, on='IDELEM')
# Antes de la predicción
future_df['IDELEM'] = pd.Categorical(future_df['IDELEM'], categories=df['IDELEM'].cat.categories)

# Predicción
predicciones = model.predict(future_df)
future_df['PREDICCION_CONTAMINACION'] = predicciones
```

```python
import matplotlib.pyplot as plt

idelem_objetivo = 3910
df_plot = future_df[future_df['IDELEM'] == idelem_objetivo]

plt.figure(figsize=(10,5))
plt.plot(df_plot['HORA'], df_plot['PREDICCION_CONTAMINACION'], marker='o')
plt.title(f'Predicción de contaminación por hora - IDELEM {idelem_objetivo}')
plt.xlabel('Hora del día')
plt.ylabel('Valor de contaminación')
plt.grid(True)
plt.show()
```

```python
unique_idelems = future_df['IDELEM'].cat.categories

for idelem in unique_idelems:
    subset = future_df[future_df['IDELEM'] == idelem]
    
    plt.figure(figsize=(8, 4))
    plt.plot(subset['HORA'], subset['PREDICCION_CONTAMINACION'], marker='o', color='steelblue')
    plt.title(f'Predicción de contaminación por hora - IDELEM {idelem}')
    plt.xlabel('Hora del día')
    plt.ylabel('Valor de contaminación')
    plt.xticks(range(0, 24))
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

# test 3

```python
# Asegúrate de tener la columna HORA en el dataframe original
df['HORA'] = df['FECHA_HORA'].dt.hour

# Obtener el valor real medio por IDELEM y hora
reales_por_hora = (
    df.groupby(['IDELEM', 'HORA'])['VALOR_CONTAMINACION']
    .mean()
    .reset_index()
    .rename(columns={'VALOR_CONTAMINACION': 'REAL_CONTAMINACION'})
)
```

```python
# Unimos valores reales y predicciones
comparacion = pd.merge(
    future_df,
    reales_por_hora,
    on=['IDELEM', 'HORA'],
    how='inner'  # usa "outer" si quieres incluir también combinaciones sin datos reales
)
```

```python
idelems = comparacion['IDELEM'].cat.categories

for idelem in idelems:
    subset = comparacion[comparacion['IDELEM'] == idelem]

    plt.figure(figsize=(8, 4))
    plt.plot(subset['HORA'], subset['REAL_CONTAMINACION'], label='Real', marker='o', color='green')
    plt.plot(subset['HORA'], subset['PREDICCION_CONTAMINACION'], label='Predicción', marker='x', color='blue')
    
    plt.title(f'Contaminación por hora - IDELEM {idelem}')
    plt.xlabel('Hora del día')
    plt.ylabel('Valor de contaminación')
    plt.xticks(range(0, 24))
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
```
