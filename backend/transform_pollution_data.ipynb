# Standarize date to later join with traffic

```python
# Create a list with hourly columns (H01 a H24)
horas = [f'H{str(i).zfill(2)}' for i in range(1, 25)]
horas
```

```python
df_largo = df.melt(
    id_vars=['ANO', 'MES', 'DIA', 'MAGNITUD', 'TECNICA', 'CODIGO', 'NOMBRE', 'COD_MAGNITUD', 'NOMBRE_MAGNITUD', 'ABREVIATURA', 'UNIDAD', 'COD_TECNICA', 'TECNICA_MEDIDA'],
    value_vars=horas,
    var_name='HORA',
    value_name='VALOR_CONTAMINACION'
)
df_largo
```

```python
df_largo['HORA'] = df_largo['HORA'].str.extract('H(\d+)').astype(int) - 1  # 0-23
# df_largo['FECHA_HORA'] = pd.to_datetime(df_largo[['ANO', 'MES', 'DIA']]) + pd.to_timedelta(df_largo['HORA'], unit='h')
df_largo['FECHA_HORA'] = pd.to_datetime(
    dict(year=df_largo['ANO'], month=df_largo['MES'], day=df_largo['DIA'])
) + pd.to_timedelta(df_largo['HORA'], unit='h')
```

```python
df_largo = df_largo.sort_values(by='FECHA_HORA').reset_index(drop=True)
df_largo
```

```python
df_largo = df_largo.dropna(subset=['VALOR_CONTAMINACION'])
df_largo

# print(df_largo[['FECHA_HORA', 'VALOR_CONTAMINACION']].head())
```

```python
df_largo = df_largo.drop(columns=['ANO', 'MES', 'DIA', 'HORA']).reset_index(drop=True)
```
